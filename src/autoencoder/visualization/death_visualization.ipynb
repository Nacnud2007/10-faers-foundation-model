```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Set publication style
sns.set_theme(style="whitegrid")
plt.rcParams.update({"font.size": 11, "axes.labelsize": 12, "axes.titlesize": 14})

# ==========================================
# DATA EXTRACTION FROM YOUR LOGS (Death ADR)
# ==========================================
epochs = np.arange(1, 8)

train_loss = [0.782355, 0.766455, 0.757764, 0.749101, 0.739525, 0.729243, 0.718234]
val_loss = [0.769818, 0.765134, 0.759716, 0.757060, 0.757694, 0.757309, 0.761811]

precision = [0.4589, 0.4463, 0.4625, 0.4690, 0.4620, 0.4639, 0.4605]
recall = [0.5156, 0.5392, 0.5274, 0.5250, 0.5372, 0.5388, 0.5450]
f1_score = [0.4856, 0.4884, 0.4928, 0.4955, 0.4968, 0.4985, 0.4992]

# ==========================================
# PLOT: TRAINING HISTORY DASHBOARD
# ==========================================
fig, axes = plt.subplots(1, 2, figsize=(14, 5.5))

# Left Panel: Loss Optimization History
axes[0].plot(epochs, train_loss, color="#1f77b4", marker="o", linewidth=2, label="Train Loss")
axes[0].plot(epochs, val_loss, color="#ff7f0e", marker="s", linewidth=2, linestyle="--", label="Val Loss")
axes[0].axvline(x=7, color="red", linestyle=":", linewidth=2, label="Best Checkpoint (Epoch 7)")
axes[0].set_title("Loss Optimization History ('Death' Target)")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Loss")
axes[0].set_ylim(0.70, 0.80)
axes[0].legend()

# Right Panel: Validation Performance Metrics (t = 0.50)
axes[1].plot(epochs, precision, color="#2ca02c", marker="^", linewidth=1.5, label="Precision")
axes[1].plot(epochs, recall, color="#d62728", marker="v", linewidth=1.5, label="Recall")
axes[1].plot(epochs, f1_score, color="#9467bd", marker="d", linewidth=2, label="F1-Score")
axes[1].axhline(y=0.2, color="#7f7f7f", linestyle="--", linewidth=2, label="Random Model F1 (0.2)")
axes[1].axvline(x=7, color="red", linestyle=":", linewidth=2, label="Best Checkpoint (Epoch 7)")
axes[1].set_title("Validation Metrics Tracking (t = 0.50)")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Score")
axes[1].set_ylim(0.10, 1.0)
axes[1].legend()

plt.tight_layout()
plt.savefig("faers_death_training_history.png", dpi=300)
plt.show()
```
